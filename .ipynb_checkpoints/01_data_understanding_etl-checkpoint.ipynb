# 01 Data Understanding and ETL

Goal: load the raw Kaggle files, understand the structure, create beginner-friendly derived features, and save cleaned train/test files for modelling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option('display.max_columns', 120)
DATA_RAW = Path('../data/raw')
DATA_PROCESSED = Path('../data/processed')
```

```python
train = pd.read_csv(DATA_RAW / 'train.csv')
test = pd.read_csv(DATA_RAW / 'test.csv')
submission = pd.read_csv(DATA_RAW / 'sample_submission.csv')

print('train shape:', train.shape)
print('test shape:', test.shape)
print('sample_submission shape:', submission.shape)
train.head()
```

## Target column

`SalePrice` exists only in `train.csv`. This is the value the model learns to predict.

```python
target = 'SalePrice'
print(train[target].describe())
```

## Missing values

Some missing values mean “not available”, but in this dataset many missing values mean the house does not have that feature, for example no garage, no pool, no fence, or no basement.

```python
missing = train.isna().sum().sort_values(ascending=False)
missing[missing > 0].head(25)
```

## Feature engineering used in ETL

We create simple features that are easy to explain in a portfolio.

```python
def add_house_features(df):
    df = df.copy()
    df['HouseAge'] = df['YrSold'] - df['YearBuilt']
    df['RemodAge'] = df['YrSold'] - df['YearRemodAdd']
    df['TotalSF'] = df['TotalBsmtSF'].fillna(0) + df['1stFlrSF'].fillna(0) + df['2ndFlrSF'].fillna(0)
    df['TotalBathrooms'] = (
        df['FullBath'].fillna(0) + 0.5 * df['HalfBath'].fillna(0) +
        df['BsmtFullBath'].fillna(0) + 0.5 * df['BsmtHalfBath'].fillna(0)
    )
    df['TotalPorchSF'] = (
        df['OpenPorchSF'].fillna(0) + df['EnclosedPorch'].fillna(0) +
        df['3SsnPorch'].fillna(0) + df['ScreenPorch'].fillna(0)
    )
    df['HasGarage'] = (df['GarageArea'].fillna(0) > 0).astype(int)
    df['HasBasement'] = (df['TotalBsmtSF'].fillna(0) > 0).astype(int)
    df['HasFireplace'] = (df['Fireplaces'].fillna(0) > 0).astype(int)
    df['HasPool'] = (df['PoolArea'].fillna(0) > 0).astype(int)
    return df

train_fe = add_house_features(train)
test_fe = add_house_features(test)

train_fe.to_csv(DATA_PROCESSED / 'train_features.csv', index=False)
test_fe.to_csv(DATA_PROCESSED / 'test_features.csv', index=False)

print('Saved processed files.')
train_fe[['SalePrice','HouseAge','RemodAge','TotalSF','TotalBathrooms','HasGarage','HasBasement']].head()
```
