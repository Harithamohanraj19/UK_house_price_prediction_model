## 1. Business Understanding

Before touching the data, we need to understand the **business problem**.

### Problem

We want to predict the final sale price of a residential house in Ames, Iowa.

### Target variable

The target variable is:

```text
SalePrice
```

This is the price that the house was sold for.

### Machine Learning problem type

This is a **supervised regression problem**.

Why?

- Supervised: because the training data includes the correct answer, `SalePrice`.
- Regression: because `SalePrice` is a continuous numerical value, not a class label.

### Possible users of this model

This kind of model could be useful for:

- House sellers who want to estimate a fair selling price.
- Buyers who want to check if a house is overpriced.
- Real estate agents who want data-driven price guidance.
- Banks or mortgage lenders estimating property value.

### Important note

A house price prediction model should not be treated as a final real estate valuation tool.  
It is an estimate based on available features in the dataset.

## 10. Data Types vs Machine Learning Types

Pandas data types are not always the same as machine learning feature types.

For example:

- `MSSubClass` is stored as a number, but it represents a dwelling type. It should be treated more like a category.
- `OverallQual` is stored as a number from 1 to 10. It has a meaningful order, so it is ordinal.
- `Neighborhood` is text and has no natural order, so it is nominal categorical.

This distinction is very important before encoding and modelling.

````python
train.dtypes.value_counts()
````

````python
numeric_columns = train.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_columns = train.select_dtypes(include=["object"]).columns.tolist()

print("Number of numeric columns:", len(numeric_columns))
print("Number of categorical/object columns:", len(categorical_columns))

print("\nNumeric columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)
````

## 11. Manual Feature Type Classification

Now we classify important features by **ML meaning**, not just pandas dtype.

This classification will help us later when we choose:

- imputation strategy
- encoding method
- scaling method
- PCA input features

````python
ml_feature_types = {
    "target": ["SalePrice"],
    "identifier": ["Id"],
    "continuous_numeric": [
        "LotFrontage", "LotArea", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
        "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF",
        "GrLivArea", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch",
        "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal"
    ],
    "count_numeric": [
        "BsmtFullBath", "BsmtHalfBath", "FullBath", "HalfBath", "BedroomAbvGr",
        "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces", "GarageCars"
    ],
    "year_numeric": [
        "YearBuilt", "YearRemodAdd", "GarageYrBlt", "YrSold", "MoSold"
    ],
    "ordinal_categorical": [
        "LotShape", "Utilities", "LandSlope", "OverallQual", "OverallCond",
        "ExterQual", "ExterCond", "BsmtQual", "BsmtCond", "BsmtExposure",
        "BsmtFinType1", "BsmtFinType2", "HeatingQC", "KitchenQual",
        "Functional", "FireplaceQu", "GarageFinish", "GarageQual", "GarageCond",
        "PavedDrive", "PoolQC", "Fence"
    ],
    "nominal_categorical": [
        "MSSubClass", "MSZoning", "Street", "Alley", "LandContour", "LotConfig",
        "Neighborhood", "Condition1", "Condition2", "BldgType", "HouseStyle",
        "RoofStyle", "RoofMatl", "Exterior1st", "Exterior2nd", "MasVnrType",
        "Foundation", "Heating", "Electrical", "GarageType", "MiscFeature",
        "SaleType", "SaleCondition"
    ],
    "binary_categorical": [
        "CentralAir"
    ]
}

summary_rows = []
for feature_type, cols in ml_feature_types.items():
    existing_cols = [col for col in cols if col in train.columns]
    summary_rows.append({
        "feature_type": feature_type,
        "number_of_columns": len(existing_cols),
        "example_columns": ", ".join(existing_cols[:8])
    })

feature_type_summary = pd.DataFrame(summary_rows)
feature_type_summary
````

### Important learning point

A column can be numeric in pandas but categorical in machine learning.

Example:

```text
MSSubClass = 20, 30, 40, 50, ...
```

These numbers are codes for house type.  
They are not real quantities.

So later we should treat `MSSubClass` as categorical, not continuous numeric.

## 12. Missing Values: Count and Meaning

Missing values are not all the same.

There are two main types:

### Type 1: Real missing values

The value should exist, but it is unknown.

Example:

```text
LotFrontage
```

If missing, we may need to estimate it using median or neighbourhood median.

### Type 2: Meaningful missing values

The missing value means the house does not have that feature.

Example:

```text
PoolQC = NA means No Pool
GarageType = NA means No Garage
FireplaceQu = NA means No Fireplace
```

This distinction is very important.

````python
missing_summary = (
    train.isna().sum()
    .reset_index()
    .rename(columns={"index": "column", 0: "missing_count"})
)
missing_summary["missing_percent"] = (missing_summary["missing_count"] / len(train) * 100).round(2)
missing_summary = missing_summary[missing_summary["missing_count"] > 0]
missing_summary = missing_summary.sort_values("missing_percent", ascending=False)
missing_summary
````

## 13. Missing Values with Business Interpretation

Below we separate columns where `NA` means "not available / not present" from columns where the value is truly unknown.

This decision comes from the data dictionary.

````python
none_meaning_columns = [
    "Alley", "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1",
    "BsmtFinType2", "FireplaceQu", "GarageType", "GarageFinish",
    "GarageQual", "GarageCond", "PoolQC", "Fence", "MiscFeature",
    "MasVnrType"
]

likely_real_missing_columns = [
    "LotFrontage", "MasVnrArea", "Electrical", "GarageYrBlt"
]

interpretation_rows = []
for col in missing_summary["column"]:
    if col in none_meaning_columns:
        meaning = "NA likely means feature not present"
        suggested_action = "Fill with 'None'"
    elif col in likely_real_missing_columns:
        meaning = "Likely real missing/unknown value"
        suggested_action = "Impute using median/mode or domain logic"
    else:
        meaning = "Needs investigation"
        suggested_action = "Check data dictionary and distributions"
    interpretation_rows.append({
        "column": col,
        "missing_count": int(train[col].isna().sum()),
        "missing_percent": round(train[col].isna().mean() * 100, 2),
        "meaning": meaning,
        "suggested_action": suggested_action
    })

missing_interpretation = pd.DataFrame(interpretation_rows)
missing_interpretation
````

### Important learning point

Do not blindly do this for every column:

```python
fillna(mean)
```

That can destroy meaningful information.

For example, if `GarageType` is missing, filling with the most common garage type is wrong.  
The correct interpretation is usually `No Garage`.

## 15. Basic Sanity Checks

Sanity checks help detect impossible or suspicious values.

Examples:

- Can `GarageArea` be negative?
- Can `YearBuilt` be after `YrSold`?
- Can `FullBath` be negative?
- Can `GrLivArea` be zero?

````python
sanity_checks = {
    "negative_lot_area": (train["LotArea"] < 0).sum(),
    "negative_living_area": (train["GrLivArea"] < 0).sum(),
    "zero_living_area": (train["GrLivArea"] == 0).sum(),
    "year_built_after_sold": (train["YearBuilt"] > train["YrSold"]).sum(),
    "remodel_after_sold": (train["YearRemodAdd"] > train["YrSold"]).sum(),
    "garage_year_after_sold": (train["GarageYrBlt"] > train["YrSold"]).sum(),
}

pd.DataFrame(sanity_checks.items(), columns=["check", "count"])
````

## 19. Transform: Basic Cleaning Decisions

This notebook should not do advanced preprocessing yet.

But we can make simple ETL decisions:

1. Keep raw data unchanged.
2. Create copies for cleaned/processed data.
3. Convert `MSSubClass` to categorical because it represents dwelling type, not quantity.
4. Save basic processed files for later notebooks.

Advanced cleaning, imputation, encoding, scaling, and PCA will happen later using pipelines.

````python
train_etl = train.copy()
test_etl = test.copy()

# MSSubClass is a numeric code, but it represents a category of dwelling.
train_etl["MSSubClass"] = train_etl["MSSubClass"].astype(str)
test_etl["MSSubClass"] = test_etl["MSSubClass"].astype(str)

print(train_etl["MSSubClass"].dtype)
print(test_etl["MSSubClass"].dtype)
````

## 20. Load: Save Processed Data

In ETL, **Load** means saving the transformed data somewhere for the next stage.

Here we save basic processed files.

Later notebooks can use these files instead of reading the raw files again.

````python
PROCESSED_DIR = Path("../data/processed")
if not PROCESSED_DIR.exists():
    PROCESSED_DIR = Path("processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

train_etl.to_csv(PROCESSED_DIR / "train_basic_processed.csv", index=False)
test_etl.to_csv(PROCESSED_DIR / "test_basic_processed.csv", index=False)

print("Saved:")
print(PROCESSED_DIR / "train_basic_processed.csv")
print(PROCESSED_DIR / "test_basic_processed.csv")
````
