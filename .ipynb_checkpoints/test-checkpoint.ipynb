I would choose the median because the dataset contains an outlier (£1.2 million), which significantly increases the mean. Since the mean is sensitive to extreme values, it does not accurately represent the typical house price. The median is not affected as much by outliers, so it provides a better measure of the typical house price in this dataset. Mode is not suitable here because house prices are mostly unique values, so there is no meaningful most frequent price.

I would use the median instead of the mean when the dataset contains outliers or is skewed. Since the mean is sensitive to extreme values, it may not represent the typical value accurately. The median is more robust because it is based on the middle value rather than the average of all values.

The mean can only be calculated for numerical data because it requires mathematical operations such as addition and division. Gender is a categorical variable, so calculating an average is not meaningful. For categorical data, the mode is usually the appropriate statistic.

I would use the mode when I want to identify the most frequently occurring value in a dataset. It is particularly useful for categorical variables such as gender, department, product category, or customer type, where calculating the mean or median is not appropriate.

The mean is calculated by dividing the sum of all values by the number of observations. Since every value contributes to the total sum, an extremely large or small value (an outlier) can significantly change the sum, causing the mean to increase or decrease. This makes the mean sensitive to outliers.

If the data contains outliers or is skewed, I would prefer the median. If the data is approximately symmetric and does not contain significant outliers, the mean is usually appropriate.

```python
import pandas as pd

houses = pd.DataFrame({
    "House_ID": ["H1","H2","H3","H4","H5","H6","H7","H8","H9","H10"],
    "House_Price": [
        165000,
        170000,
        180000,
        220000,
        250000,
        260000,
        310000,
        330000,
        450000,
        1200000
    ]
})

houses
```

```python
Q1 = houses["House_Price"].quantile(0.25)

Q3 = houses["House_Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

print(lower)

print(upper)
```

```python
houses[(houses['House_Price']< lower) | (houses['House_Price'] > upper)]
```

. Why do Data Scientists calculate IQR?
ata Scientists calculate the Interquartile Range (IQR) to measure the spread of the middle 50% of the data and to identify potential outliers. Since IQR focuses on the central portion of the dataset, it is less affected by extreme values than the range.

Why don't we use Mean to detect Outliers?
- We do not use the mean to detect outliers because the mean is influenced by extreme values. An outlier changes the mean itself, making it an unreliable measure for identifying unusual observations. Instead, IQR is preferred because it is based on the middle 50% of the data and is much less affected by extreme values.

Which statistic tells us which dataset is more spread out?
Variance and Standard Deviation measure how spread out the data points are from the mean. They help us understand the variability in the dataset, even when two datasets have the same average.

Why do we square the differences when calculating variance?
We square the differences to make all values positive so that positive and negative differences do not cancel each other out. Squaring also gives more importance to observations that are far away from the mean.

Why do we use Standard Deviation instead of Variance when explaining results to business users?
Variance is measured in squared units (for example, marks² or pounds²), which are difficult to interpret. Standard Deviation is the square root of the variance, so it is expressed in the original unit of the data, making it much easier for business users to understand.

Why do Data Scientists use histograms?
Data Scientists use histograms to visualize the distribution of numerical data. Histograms help identify the spread of the data, skewness, outliers, and whether the data follows a normal distribution.

What does the height of a histogram bar represent?
the frequency (count) of observations within a particular range (bin).

A right-skewed distribution has most observations concentrated on the left, with a long tail extending to the right due to a few large values (outliers).

When would you apply a log transformation?
- I would apply a log transformation when the numerical data is heavily right-skewed or contains very large values that create skewness. Log transformation compresses large values more than small values, reducing skewness and making the distribution closer to normal.

- Log transformation reduces skewness by compressing large values more than smaller values. This reduces the influence of extreme values, making the distribution more symmetric and often improving machine learning model performance.

Correlation measures the strength and direction of the relationship between two variables.

--In a positive correlation, both variables move in the same direction. As one variable increases, the other also tends to increase. In a negative correlation, the variables move in opposite directions. As one variable increases, the other tends to decrease.

A correlation of 0 means there is no linear relationship between the two variables. Changes in one variable do not show a consistent increase or decrease in the other.

Correlation does not imply causation.
- No. Correlation only measures the relationship between two variables. It does not prove that one variable causes the other. There may be other factors influencing both variables.

- No. A low correlation does not automatically mean a feature is useless. The feature may have a non-linear relationship with the target, may become useful when combined with other features, or may be important for certain machine learning models. We should investigate further before removing it.

Feature scaling transforms numerical features onto a similar scale so that features with larger values do not dominate features with smaller values during model training.

Decision Trees split data using conditions such as "Feature > Threshold". Since scaling changes only the magnitude of values and not their order, the split decisions remain the same. Therefore, Decision Trees do not require feature scaling.

PCA should be performed after scaling because PCA is based on variance. If features have different scales, the feature with the largest variance will dominate the principal components. Scaling ensures that every feature contributes fairly to PCA.

StandardScaler applies the same transformation to every observation. It changes the scale of the values but preserves their relative order. Therefore, the smallest value remains the smallest and the largest value remains the largest.

Scaling is important before algorithms like KNN and PCA because they depend on distances or variances. Without scaling, features with larger numerical values dominate the calculations, causing the model to give them more importance than smaller-scale features.

```python
import pandas as pd

from sklearn.linear_model import LinearRegression

from sklearn.model_selection import cross_val_score

data = {
    "Size":[1000,1200,1500,1800,2000],
    "Price":[200000,230000,290000,340000,380000]
}

df = pd.DataFrame(data)

X = df[["Size"]]

y = df["Price"]

model = LinearRegression()

scores = cross_val_score(
    model,
    X,
    y,
    cv=5
)

print(scores)

print(scores.mean())
```
